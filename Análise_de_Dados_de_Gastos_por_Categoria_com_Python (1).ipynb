## Objetivo
Este projeto tem como objetivo analisar dados de gastos por categoria utilizando Python, explorando técnicas de manipulação de dados e visualização.

Nesta etapa, os dados são estruturados em formato de tabela, utilizando a biblioteca pandas, permitindo sua organização e posterior análise.

```python
import pandas as pd

dados = {"Categoria": ["Alimentação", "Transporte", "Lazer", "Alimentação", "Transporte"],"Valor": [50, 20, 100, 80, 30]}

df = pd.DataFrame(dados)

print(df)
```

Aqui são realizadas operações básicas, como cálculo do valor total e média dos gastos, permitindo uma visão geral dos dados.

```python
print("Total gasto:", df["Valor"].sum())
print("Média:", df["Valor"].mean())
```

Os dados são agrupados por categoria, possibilitando identificar padrões e comparar os gastos entre diferentes grupos.

```python
df.groupby("Categoria")["Valor"].sum()
```

Nesta etapa, é gerado um gráfico de barras para representar visualmente os gastos por categoria, facilitando a interpretação dos resultados.

```python
import matplotlib.pyplot as plt

df.groupby("Categoria")["Valor"].sum().plot(kind="bar")
plt.title("Gastos por Categoria")
plt.show()
```

A análise permitiu compreender a distribuição dos gastos entre as categorias, evidenciando como a utilização de ferramentas de análise de dados pode auxiliar na interpretação de informações e na tomada de decisões.
